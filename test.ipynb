```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
# Load the data
df_sst = pd.read_csv('dataset/sst.csv', skiprows=[1])
df_dhw = pd.read_csv('dataset/dhw.csv', skiprows=[1])
```

```python
df_sst.head()
```

```python
df_dhw.head()
```

```python
# Merge and Clean
df = pd.merge(df_sst, df_dhw, on=['time', 'latitude', 'longitude'])
df['time'] = pd.to_datetime(df['time'])
```

```python
df.head()
```

```python
# Convert time to "Days since start"
df['days'] = (df['time'] - df['time'].min()).dt.days
```

```python
# Normalize the data (Neural Networks love numbers between 0 and 1)
scaler_temp = MinMaxScaler()
scaler_time = MinMaxScaler()

t_data = scaler_time.fit_transform(df[['days']])
temp_data = scaler_temp.fit_transform(df[['analysed_sst']])
```
