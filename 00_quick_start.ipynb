## 💡 Understanding PINNs

### What makes this a Physics-Informed Neural Network?

Traditional neural networks only learn from data. **PINNs** incorporate physical laws:

1. **Data Loss**: 
   - Minimize prediction error at sensor locations
   - Ensures accuracy where we have measurements

2. **Physics Loss**:
   - Enforces smooth temperature gradients
   - Based on heat diffusion principles
   - Ensures predictions obey physical laws

**Total Loss = Data Loss + λ × Physics Loss**

This makes predictions both **accurate** and **physically realistic**!
